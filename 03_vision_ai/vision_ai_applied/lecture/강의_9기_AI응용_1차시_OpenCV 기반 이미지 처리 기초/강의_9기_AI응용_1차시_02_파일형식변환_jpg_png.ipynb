```python
from PIL import Image

path = '/content/idol_girl.png'

img_idol = Image.open(path)
img_idol
```

```python
# RGBA >> RGB 로 변환 (JPG 투명도 지원 없음)
if img_idol.mode == "RGBA": # A : 투명도
  img_idol = img_idol.convert("RGB")
```

```python
img_idol.save("img_idol.jpg", "JPEG", quality=95)
```

```python
# 폴더 전체 jpg >> png

import os
from PIL import Image

input_folder = "images"
output_folder = "output"

os.makedirs(input_folder, exist_ok=True)
os.makedirs(output_folder, exist_ok=True)

for file in os.listdir(input_folder):
    if file.endswith('.jpg'):
        img = Image.open(os.path.join(input_folder, file))

        new_name = file.replace('.jpg', '.png')
        img.save(os.path.join(output_folder, new_name), "PNG")
```

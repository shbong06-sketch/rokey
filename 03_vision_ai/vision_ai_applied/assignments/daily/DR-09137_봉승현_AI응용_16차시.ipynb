## [문제 1]

YOLOv10은 기존 YOLO 시리즈와 달리 추론 속도를 늦추는 후처리 단계를 제거하기 위해 'One-to-One Matching' 전략을 도입하였습니다. 기존의 중복된 바운딩 박스를 제거하는 후처리 기법의 명칭은 무엇이며, YOLOv10이 이를 제거함으로써 얻는 주요 이점은 무엇입니까?

**[답안 작성란]:**
NMS(Non-Maximum Suppression)

기존 NMS 방식은 하이퍼파라미터(IoU 임계값)에 따라 검출 결과가 크게 달라지고, CPU 병렬 연산 후 별도의 순차적 연산이 필요해 전체 속도가 저하되는 문제가 존재하였다. YOLOv10은 NMS-Free 방식을 채택하여 학습 단계에서 중복을 스스로 억제하도록 훈련시켜, 추론 시에는 후처리가 필요 없는 깔끔한 결과만 출력하도록 하여 진정한 의미의 end-to-end 검출을 실현했다.

## CNN 예제 - nn.Conv2d

```python
import torch

cnn1 = torch.nn.Conv2d(2, 3, kernel_size=4, stride=1, padding=1)

'''
* Conv2d arguments: (input_channel, output_channel, kernel_size, stride, padding)
* 그 외 인자: dilation=1, groups=1, bias=True, padding_mode='zeros', device=None, dtype=None
* 자세한 사항은 https://docs.pytorch.org/docs/stable/generated/torch.nn.Conv2d.html 참고
'''

# cnn의 weight와 bias 선언
cnn1.weight.data = torch.randint(low=1, high=5, size=(3, 2, 2, 2), dtype=torch.int).float() # cnn의 weight shape은 (output_dim, input_dim, kernel_size, kernel_size) 순임.
cnn1.bias.data = torch.tensor([1,2,3]).float() # cnn의 bias shape은 output_dim 이랑 같음.

# 무작위 입력 생성
input = torch.randint(low=0, high=3, size=(2, 5, 5), dtype=torch.int).float()
print("입력 텐서: ", input)

# TO-DO: 입력의 shape 확인하기
print(input.shape)

print("=" * 100)

# TO-DO: cnn의 weight와 bias 확인하기
print(cnn1.weight)
print(cnn1.bias)

print("=" * 100)

# TO-DO: cnn의 weight와 bias의 shape 확인하기
print(cnn1.weight.shape)
print(cnn1.bias.shape)

print("=" * 100)

# TO-DO: cnn1 layer의 출력과 그 shape 확인하기
output = cnn1(input)
print(output)
print(output.shape)

print("=" * 100)
```

## CNN의 장점

Convolutional Layer(CNN 레이어)는 Fully Connected(FC) Layer에 비해 몇 가지 중요한 장점이 있습니다. 이러한 장점은 CNN이 특히 이미지 데이터에서 뛰어난 성능을 발휘하게 하는 요인입니다.

### 1. 공간적 관계 유지
> **FC Layer**: 모든 입력 뉴런이 모든 출력 뉴런에 연결되므로 입력의 공간적 구조가 무시됩니다. 따라서 FC Layer는 입력 데이터의 공간적 관계를 인식할 수 없습니다.

> **Convolution Layer**: 입력 이미지의 작은 영역(리셉티브 필드)에 대해서만 필터를 적용합니다. 이로 인해 CNN은 인접 픽셀 간의 관계를 유지하고 학습할 수 있으며, 이미지의 공간적 패턴을 인식하는 데 유리합니다.

### 2. 파라미터 수 감소
> **FC Layer**: 입력과 출력의 모든 뉴런이 연결되므로, 입력 크기와 출력 크기에 따라 파라미터 수가 급격히 증가합니다.

> **Convolution Layer**: 동일한 필터를 이미지 전체에 걸쳐 적용하므로, 파라미터 수가 크게 줄어듭니다. 이는 파라미터 공유(parameter sharing)로 인해 가능하며, 모델이 더 효율적이고 메모리 절약적입니다.

### 3. 변환 불변성(Translation Invariance)
> **FC Layer**: 입력 데이터의 위치 변화에 민감하여, 데이터의 이동이나 왜곡에 취약합니다.

> **Convolution Layer**: 필터가 이미지의 모든 위치에 동일하게 적용되기 때문에, 객체의 위치 변화에 대해 비교적 강인합니다. 이는 이미지 내 객체의 위치가 약간 변해도 그 특징을 잘 인식할 수 있게 합니다.

# Pooling
* 풀링의 목적: 특징 맵의 크기 감소, 계산 효율성 증대, 오버피팅 방지

## Max Pooling
* 특징맵에서 가장 두드러지는 특징을 찾도록 한다.

```python
input = torch.tensor([[[12 , 20 , 30, 0 ],
                       [8  , 12 , 2 , 0 ],
                       [34 , 70 , 37, 4 ],
                       [112, 100, 25, 12]]])

MaxPool = torch.nn.MaxPool2d(kernel_size=2, stride=2, padding=0)
Output = MaxPool(input)
print(Output)
```

## Average Pooling
* 전체적인 정보를 보존하면서 특징맵 크기 줄일 때 사용

```python
input = torch.tensor([[[12 , 20 , 30, 0 ],
                       [8  , 12 , 2 , 0 ],
                       [34 , 70 , 37, 4 ],
                       [112, 100, 25, 12]]])

AvgPool = torch.nn.AvgPool2d(kernel_size=2, stride=2, padding=0)
Output = AvgPool(input)
print(Output)
```

# CNN 예제 - MNIST 데이터 학습

## MNIST

![image](https://github.com/user-attachments/assets/0873d1cb-0645-4598-9ba1-d39d2e35953e)

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader

# dataset 전처리
transform = transforms.Compose([
    transforms.ToTensor(),  # 이미지를 Tensor로 변환
    transforms.Normalize((0.5,), (0.5,))  # 정규화: 평균 0.5, 표준편차 0.5로 스케일링
])

# MNIST dataset 다운로드 및 로드
train_dataset = torchvision.datasets.MNIST(root='./data', train=True, download=True, transform=transform)
test_dataset = torchvision.datasets.MNIST(root='./data', train=False, download=False, transform=transform)
```

```python
# 훈련 데이터셋의 첫번째 데이터 정보 출력
train_dataset[0]
```

```python
# 첫번째 훈련 데이터의 이미지 shape 출력
train_dataset[0][0].shape
```

```python
# DataLoader를 사용하여 데이터 배치 처리
train_loader = DataLoader(train_dataset, batch_size=128, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=128, shuffle=False)
```

```python
# CNN 모델 정의
class SimpleCNN(nn.Module):
    def __init__(self):
        super().__init__()
        # 첫 번째 Convolution layer: 입력 채널 1개, 출력 채널 32개, 커널 크기 3x3, stride=1, padding=1
        self.conv1 = nn.Conv2d(1, 32, kernel_size=3, stride=1, padding=1)
        # MaxPooling layer: 2x2 풀링
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2, padding=0)
        # 두 번째 Convolution layer: 입력 채널 32개, 출력 채널 64개, 커널 크기 3x3
        self.conv2 = nn.Conv2d(32, 64, kernel_size=3, stride=1, padding=1)
        # fc-lyaer 1: 64채널 7x7 이미지 -> 128개의 출력
        self.fc1 = nn.Linear(64 * 7 * 7, 128)
        # fc-lyaer 2: 128개의 출력 -> 10개의 클래스 (0-9 숫자)
        self.fc2 = nn.Linear(128, 10)
        # nn.ReLU() 활성화 함수를 self.relu로 저장
        self.relu = nn.ReLU(inplace=True)
        # 벡터로 평탄화 시키는 nn.Flatten()을 self.flatten으로 저장
        self.flatten = nn.Flatten()

        self.features = nn.Sequential(
            self.conv1, self.relu, self.pool,
            self.conv2, self.relu, self.pool
        )

        self.classifier = nn.Sequential(
            self.fc1, self.relu, self.fc2
        )

    def forward(self, x):
        x = self.features(x)
        x = self.flatten(x)
        x = self.classifier(x)
        # # TO-DO: 첫 번째 convolution layer -> ReLU -> Max pooling
        # x = self.conv1(x)
        # x = self.relu(x)
        # x = self.pool(x)    # 크기: [8, 32, 14, 14]

        # # TO-DO: 두 번째 convolution layer -> ReLU -> Max pooling
        # x = self.conv2(x)
        # x = self.relu(x)
        # x = self.pool(x)    # 크기: [8, 64, 7, 7]

        # # TO-DO: view 함수로 특징맵을 1차원으로 펼치기(위 init에서 self.flatten도 사용 가능)
        # x = self.flatten(x) # 크기: [8, 64 * 7 * 7]

        # # TO-DO: 첫 번째 fc-layer -> ReLU
        # x = self.fc1(x)
        # x = self.relu(x)    # 크기: [8, 128]

        # # TO-DO: 두 번째 fc-layer -> 최종 logit 출력
        # x = self.fc2(x)     # 크기: [8, 10]

        return x

# device 저장. gpu 사용 가능하면 gpu로 설정.
device = ("cuda:0" if torch.cuda.is_available() else 'cpu')

# 모델 인스턴스 생성
model = SimpleCNN().to(device)

# 손실 함수 및 옵티마이저 정의
criterion = nn.CrossEntropyLoss()  # 교차 엔트로피 손실 함수
optimizer = optim.Adam(model.parameters(), lr=0.001)  # Adam 옵티마이저 사용, 학습률 0.001

# Epoch 수 결정
epochs = 10
```

```python
# 모델 정보 summary 출력
import torchsummary

torchsummary.summary(model,(1, 28, 28))
```

```python
# 모델 훈련 함수
def train_model(model, train_loader, criterion, optimizer, num_epochs):
    # 지정된 epoch 수만큼 반복
    for epoch in range(num_epochs):
        running_loss = 0.0
        # 데이터 배치 단위로 반복
        for images, labels in train_loader:
            images, labels = images.to(device), labels.to(device) # 입력과 정답의 device를 gpu로 변환
            optimizer.zero_grad()  # optimizer의 기울기 초기화
            outputs = model(images)  # 모델의 예측값 계산 (Batch_size, 10)
            loss = criterion(outputs, labels)  # 손실 계산
            loss.backward()  # backpropagation을 통해 기울기 계산
            optimizer.step()  # 가중치 업데이트

            running_loss += loss.item()

        # epoch당 평균 손실 출력
        print(f'Epoch [{epoch+1}/{num_epochs}], Loss: {running_loss/len(train_loader):.4f}')

# 모델 평가 함수 정의
def evaluate_model(model, test_loader):
    model.eval()
    correct = 0
    total = 0
    with torch.no_grad():
        for images, labels in test_loader:
            images, labels = images.to(device), labels.to(device) # 입력과 정답의 device를 gpu로 변환
            outputs = model(images) # 모델의 예측값 계산
            _, predicted = torch.max(outputs.data, 1) # 예측한 결과 출력
            total += labels.size(0) # total: test한 총 데이터 개수
            correct += (predicted == labels).sum().item() # 모델이 맞춘 개수 계산

    print(f'\n테스트 정확도: {100 * correct / total:.2f}%')


train_model(model, train_loader, criterion, optimizer, epochs) # 모델 학습시키는 함수 가동하여 모델 학습

evaluate_model(model, test_loader) # 학습된 모델을 테스트하기 위해 평가 함수 실행
```

```python
class Country:
    name = '미입력'
    population = '미입력'

    def show(self):
        print("부모 클래스 출력입니다.")

class Korea(Country):
    def __init__(self, name, population):
        self.name = name
        self.population = population

    def show_name(self):
        print(f"여기는 {self.name} 입니다.")

    def show_pop(self):
        print(f"여기는 {self.population} 명 입니다.")

a = Korea("대한민국", '5000만명')
a.show()
a.show_name()
a.show_pop()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x,a):
    return (1/(1+np.exp(-a*x)))

x = np.arange(-2, 2.1, 0.25)
y = sigmoid(x,1)

plt.plot(x,y, c='r')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Sigmoid Function Graph')
plt.grid()
plt.show()
```

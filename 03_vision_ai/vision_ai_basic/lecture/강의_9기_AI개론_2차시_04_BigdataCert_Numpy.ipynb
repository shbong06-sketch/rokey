```python
import numpy as np

arr1 = np.array([1,2,3,4,5])
arr2 = np.array([[1,2,3],[4,5,6]])

print(arr1)
print(arr2)
print()
print(arr1.shape)   # 1차원 (5,)
print(arr2.shape)   # 2차원 (2,3)
```

```python
# arange()와 reshape()  ***중요
# arange() : array+range
# arange(start, end, step)

arr7 = np.arange(20, 200, 10)

print(arr7)
print(arr7.shape)
print(len(arr7))
```

```python
# 반드시 알아야 해!! (변환)
arr7.reshape(3, -1)
```

```python
arr8 = arr7.reshape(3, -1)
arr8
```

```python
# 배열의 산술연산: 배열의 연산은 배열 원소에 각각 적용된다

ar1 = np.arange(1,11,1)
ar2 = ar1 + 3   # broadcasting(각 원소에 적용)
ar3 = ar1 * 3

print(ar1)
print()
print(ar2)
print()
print(ar3)
print()
```

```python
# 배열의 통계 메소드 이용
arr = np.array([[5,7,9],[-7,-6,19],[6,9,11]])

print(arr)
print(arr.sum())    # 전체 요소의 합
print(arr.mean())
print(arr.max())
print(arr.min())
print(arr.argmax()) # 배열에서 가장 큰 값을 가진 인덱스
```

```python
print(arr.max(axis=0))
print(arr.max(axis=1))
```

```python
# np.where
# -> SQL에서 유래(Select/From/Where)
# -> Where: 조건

# arr < 0 조건식
# 조건식이 참(True) (arr < 0) >>> 0 반환, 아니면 arr 출력

np.where(arr<0, 0, arr)
```

```python
# np.sort() vs arr.sort() 차이 : 원본 수정 여부

import numpy as np

arr10 = np.array([3, 1, 4, 5, 2, 1, 5, 3, 4])
print(arr10)
```

```python
np.sort(arr10)  # 원본 변경 없음 (default: 오름차순)
```

```python
arr10
```

```python
arr10.sort()    # 원본 변경됨 (default: 오름차순)
arr10
```

```python
# argsort() : 정렬된 인덱스
arr11 = np.array([30, 10, 40, 20])
print(arr11)
```

```python
np.argsort(arr11)
# 순서 맞춰줄 때 사용
```

```python
import rclpy
from std_msgs.msg import String

rclpy.init()
node = rclpy.create_node('my_node')
# 주의 : 두번 누르면 RuntimeError 난다.
```

```python
# 터미널 명령어 사용 방법 -> !
!ros2 node list
```

```python
pub = node.create_publisher(String, 'chatter', 10)
m = String()
m.data = "Hello Ros2"
pub.publish(m)
```

```python
node.create_subscription(String, "chatter", lambda msg: print("받음: ", msg.data), 10)
rclpy.spin_once(node, timeout_sec=1.0)
```

```python
node.destroy_node()
rclpy.shutdown()
```

### turtlesim의 거북이를 teleop key를 이용해서 동작시키면 전/후진은 2만큼, 회전도 2만큼 값이 전달이 된다.
- 2만큼은 너무 값이 커서 정밀 컨트롤이 안된다.
- 그래서 2 -> 1만큼 움직이도록 pub을 구성해서 동작시킨다.

### 체크 관점
- ros2 topic을 이용해서 대상(거북이, 자율주행로봇, 로봇팔, 안드로이드 로봇)을 움직이는 원리를 이해

```python
import rclpy
from geometry_msgs.msg import Twist

rclpy.init()
node = rclpy.create_node('move_node')
```

```python
!ros2 topic list
```

```python
!ros2 topic info /turtle1/cmd_vel
```

```python
!ros2 interface show geometry_msgs/msg/Twist
```

```python
!ros2 service call /reset std_srvs/srv/Empty {}
```

```python
pub = node.create_publisher(Twist, 'turtle1/cmd_vel', 10)
m = Twist()
m.linear.x = 1.0
pub.publish(m)
```

```python
node.destroy_node()
rclpy.shutdown()
```

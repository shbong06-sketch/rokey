```python
# 초기화 및 노드 생성
import rclpy
from std_msgs.msg import String

rclpy.init()
node = rclpy.create_node("nb")
```

```python
# publisher
pub = node.create_publisher(String, "chatter", 10) # slef.pub = self.create_publisher(String, "chatter", 10)
m = String()
m.data = "hello1"
pub.publish(m)
```

```python
# 구독자
def my_callback(msg):
    print(msg.data)

node.create_subscription(String, 'chatter', my_callback, 10)
# 한번씩 동작하게 하는 스핀 적용, 안전 장치로 timeout 지정
rclpy.spin_once(node, timeout_sec=1.0)
```

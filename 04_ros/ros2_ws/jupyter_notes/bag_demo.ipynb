```python
from rosbag2_py import SequentialReader, StorageOptions, ConverterOptions
from rclpy.serialization import deserialize_message
# serialization, 직렬화 -> 네트워크를 통해 자료를 주고 받을 때, 데이터의 형태를 네트워크에서 보낼 수 있는 형태로 변환한다. ROS의 역할(통신 담당)
from geometry_msgs.msg import Twist
```

```python
# ros2 bag record -o turtle_run2 /cmd_vel
# 경로 개념
```

```python
reader = SequentialReader()
reader.open(
    StorageOptions(uri= '/home/shbong/worksapce/rokey/rokey/04_ros/turtle_run2', storage_id='mcap'),
    ConverterOptions('','')
)
# reader(제너레이터)에서 값을 읽고, 기타 처리
```

```python
output = reader.read_next()
```

```python
output
'''
('/turtle1/pose',   # 토픽명
 b'\x00\x01\x00\x00es\xe8@\xe7\x8a\xb2@\x00\x00\x80\xa5\x00\x00\x00\x00\x00\x00\x00\x00',   # 직렬화된 데이터
 1784598720094413264)   # 시간
'''
# reader에서 read_next 하면, (토픽, 데이터, 시간)을 반환
```

```python
'''
while 다음 내용이 있는가? :
    내용 읽기
    처리
'''
speeds = []
while reader.has_next():
    topic, data, t = reader.read_next()
    if topic == '/turtle1/cmd_vel':
        # print(f'topic : {topic} | data : {data} | time : {t}')
        msg = deserialize_message(data, Twist)  # Twist >> msg.linear.x ...
        speeds.append(msg.linear.x)
```

```python
# 값이 몇 개인가?
len(speeds)
```

```python
# 평균은 얼마인가?
sum(speeds)/len(speeds)
```
